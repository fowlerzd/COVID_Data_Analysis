```python
import pandas as pd
import plotly.express as px
```

```python
# Download the data set
import kagglehub

path = kagglehub.dataset_download("imdevskp/corona-virus-report")

print("Path to dataset files:", path)
```

```python
# Open the dataset
covid_data = pd.read_csv(path + "/covid_19_clean_complete.csv")

display(covid_data.tail())

# Find the number of unique countries
print("Num Unique Countries:", len(pd.unique(covid_data["Country/Region"])))

# Find the range of dates for the data set
dates = pd.to_datetime(covid_data["Date"]) # Convert dates to pd.datetime
print("From:", min(dates))
print("To:", max(dates))

# Add month for use in animation later
covid_data["Date"] = pd.to_datetime(covid_data["Date"])
covid_data["Month"] = covid_data["Date"].dt.month
covid_data["Year"] = covid_data["Date"].dt.year
```

```python
# Create animation of Active cases by Month
fig = px.scatter_map(covid_data,
                       lat="Lat",
                       lon="Long",
                       hover_name="Country/Region",
                       size="Confirmed",
                       color="WHO Region",
                       animation_frame="Month",
                       zoom=0.5,
                       title=f"Confirmed Covid Cases {min(dates).strftime('%m/%d/%Y')} to {max(dates).strftime('%m/%d/%Y')} [m/d/y]"
                    )
fig.show()
```
